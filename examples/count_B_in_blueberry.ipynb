```python
import numpy as np
from sand_bob import generate_code, config_kiara

config_kiara()
```

```python
ground_truth = 2
```

```python
results = generate_code("""
Count the number of b's in blueberry.
The expected result is the number only.
""", 
              n_parallel=3,
              n_iterative=2,
              n_codefix_attempts=2,
              n_feedback_iterations=1,
              final_touch=True,
                       )

results
```

```python
results._render_former_results_overview_table()
```

```python
[r.final_result for r in results  if r is not None]
```

```python
numeric_results = [int(r.final_result) for r in results if r.final_result is not None]
numeric_results
```

```python
# Count correct guesses
count_correct = np.sum(np.array(numeric_results) == ground_truth)

print(f"Ground truth: {ground_truth}")
print(f"Correct counts: {count_correct} out of {len(results)}")
print(f"Percentage correct: {count_correct/len(results)*100:.1f}%")
```
