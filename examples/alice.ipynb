# The `%%alice` magic command
The `%%alice` magic command allows to generate code and store it a Jupyter Notebooks without a lot of boilerplate code.
It is recommended though to initialize the environment, e.g. to define folders where alice should have access to or how many notebooks should be generated in parallel.  

**Note:** The GUI will not be rendered on Github, you need to execute this notebook to see it.

```python
from sand_bob import initialize, config_scadsai_llm
config_scadsai_llm()

initialize(input_host_path="input_data/")
```

```python
%%alice
Plot histograms of all columns in input_data/measurements.csv
```

```python
%%alice
Visualize a coloured correlation matrix of all columns in input_data/measurements.csv
```

By initializing multiple parallel attempts, you can also run analysis in parallel and inspect results interactively.

```python
initialize(input_host_path="input_data/", n_parallel=5)
```
