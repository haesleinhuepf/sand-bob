```python
import numpy as np
import matplotlib.pyplot as plt

# Generate data for the sigmoid
x = np.linspace(-10, 10, 400)
y = 1 / (1 + np.exp(-x))

# Plot the sigmoid
plt.figure(figsize=(6, 4))
plt.plot(x, y, label='Sigmoid')
plt.title('Sigmoid Function')
plt.xlabel('x')
plt.ylabel('σ(x)')
plt.grid(True)
plt.legend()

# Second‑last output: description of what was done
print("Sigmoid curve plotted.")

# Last output: display the plot
plt.show()
```
